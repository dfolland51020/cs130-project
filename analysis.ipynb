```python
import pandas as pd
import numpy as np #library

state_crime = pd.read_csv('state_crime.csv')

#only using the columns that I am comparing in my question
columns_needed = [
    'Data.Totals.Property.Burglary',
    'Data.Totals.Property.Larceny',
    'Data.Totals.Property.Motor',
    'Data.Totals.Violent.Assault',
    'Data.Totals.Violent.Murder',
    'Data.Totals.Violent.Rape',
    'Data.Totals.Violent.Robbery'
]

#locating the states that are only in the year 2000 and only finding them in the columns_needed (.loc)
state_crime_2000 = state_crime.loc[
    state_crime['Year'] == 2000,
    ['State'] + ['Year'] + columns_needed
]

#delete United States and District of Columbia
#the .isin I looked up on the pandas website; this keeps the list of states that is NOT (~) in the list of excluded_areas
exluded_areas = ['United States', 'District of Columbia']
state_crime_2000 = state_crime_2000[~state_crime_2000['State'].isin(exluded_areas)]

#comparing each element in the row and producing the highest number and finding that column
#adding new columns to state_crime.csv
#gives the column name of the maximum value for each row; need 1 as if you use 0, it returns NaN; looked up on pandas
state_crime_2000['Highest_Crime'] = state_crime_2000[columns_needed].idxmax(axis=1)

#gives the maximum value in that column; need 1 as if you use 0, it returns NaN; looked up on pandas
state_crime_2000['Highest_Value'] = state_crime_2000[columns_needed].max(axis=1)

#prints the values that I want; state, crime, value
#to_string(index=False) deletes the indexing number; looked up on pandas
print(state_crime_2000[['State', 'Highest_Crime', 'Highest_Value']].to_string(index=False))
```

```python
import pandas as pd
import numpy as np #library

state_crime = pd.read_csv('state_crime.csv')

#only using the columns that I am comparing in my question
columns_needed = [
    'Data.Totals.Property.Burglary',
    'Data.Totals.Property.Larceny',
    'Data.Totals.Property.Motor',
    'Data.Totals.Violent.Assault',
    'Data.Totals.Violent.Murder',
    'Data.Totals.Violent.Rape',
    'Data.Totals.Violent.Robbery'
]

#locating the states that are only in the year 2000 and only finding them in the columns_needed (.loc)
state_crime_2000 = state_crime.loc[
    state_crime['Year'] == 2000,
    ['State'] + ['Year'] + columns_needed
]

#delete United States and District of Columbia
#the .isin I looked up on the pandas website; this keeps the list of states that is NOT (~) in the list of excluded_areas
exluded_areas = ['United States', 'District of Columbia']
state_crime_2000 = state_crime_2000[~state_crime_2000['State'].isin(exluded_areas)]

#comparing each element in the row and producing the highest number and finding that column
#adding new columns to state_crime.csv
#gives the column name of the maximum value for each row; need 1 as if you use 0, it returns NaN; looked up on pandas
state_crime_2000['Highest_Crime'] = state_crime_2000[columns_needed].idxmax(axis=1)

#gives the maximum value in that column; need 1 as if you use 0, it returns NaN; looked up on pandas
state_crime_2000['Highest_Value'] = state_crime_2000[columns_needed].max(axis=1)

#want to now find the highest value of all in the higest value in this list
#need axis 0 becuase you are only looking at the maximum in 1 column; looked up on pandas
highest_of_all = state_crime_2000['Highest_Value'].idxmax(axis=0)

#now that we have found it, in the column, we need to find the row; looked up on pandas
highest_of_all_state = state_crime_2000.loc[highest_of_all, ['State', 'Highest_Crime', 'Highest_Value']]

#prints the values that I want; state, crime
#to_string(index=False) deletes the indexing number; looked up on pandas
print(highest_of_all_state.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

#sort by violet rates
plot_data = state_crime_2000.sort_values('State')

#making the graph more easily readable
plt.figure(figsize=(12, 8))

#Create bar graph
plt.barh(plot_data['State'], plot_data['Highest_Value'])

plt.xlabel('Highest Crime Value')
plt.ylabel('State')
plt.title('Highest Crime Value by State (Year 2000)')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load data
state_crime = pd.read_csv('state_crime.csv')

# Filter for year 2000
state_crime_2000 = state_crime[state_crime['Year'] == 2000].copy()

# Remove undesired areas
excluded_areas = ['United States', 'District of Columbia']
state_crime_2000 = state_crime_2000[~state_crime_2000['State'].isin(excluded_areas)]

# Select only the columns needed
columns_graph1 = ['State', 'Data.Rates.Violent.All', 'Data.Rates.Property.All']
plot_data = state_crime_2000[columns_graph1].set_index('State')

# Create bar graph
plt.figure(figsize=(14, 8))

# Create side-by-side bars
x = np.arange(len(plot_data))  # positions for states
width = 0.4                     # bar width

plt.bar(x - width/2, plot_data['Data.Rates.Violent.All'], 
        width, label='Violent Crime Rate', color='red')
plt.bar(x + width/2, plot_data['Data.Rates.Property.All'], 
        width, label='Property Crime Rate', color = 'green')

# Labeling
plt.xticks(x, plot_data.index, rotation=90)
plt.ylabel('Crime Rate')
plt.title('Violent vs Property Crime Rates by State (2000)')
plt.legend()

plt.tight_layout()
plt.show()

#I googled how to do this one as I thought it would look cool but this graph doesn't count because I googled it
```
